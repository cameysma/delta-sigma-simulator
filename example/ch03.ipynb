```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 10000

V = 0.2

n = np.arange(1, N + 1)[:, np.newaxis]

f0 = 1

t = np.arange(-1 / (2 * f0), 1 / (2 * f0), 1 / (1000 * f0))

w = 4 / np.pi * np.sum(1 / n * np.sin(np.pi * n / 2) * np.cos(2 * np.pi * n * f0 * t), axis=0)
```

```python
y_0 = np.sum(1 / n ** 2 * np.sin(np.pi * n * (V + 1) / 2) * np.sin(2 * np.pi * n * f0 * t), axis=0)

y_1 = np.zeros_like(y_0)
y_1 += np.sum(1 / n ** 2 * np.sin(np.pi * n * V / 2) * np.cos(n * np.pi / 2) * np.sin(2 * np.pi * n * f0 * t), axis=0)
y_1 += np.sum(1 / n ** 2 * np.cos(np.pi * n * V / 2) * np.sin(n * np.pi / 2) * np.sin(2 * np.pi * n * f0 * t), axis=0)

y_2 = np.zeros_like(y_0)
y_2 += +1 / 2 * np.sum(1 / n ** 2 * np.cos(n * np.pi / 2) * np.cos(2 * np.pi * n * f0 * (t - V / (4 * f0))), axis=0)
y_2 += -1 / 2 * np.sum(1 / n ** 2 * np.cos(n * np.pi / 2) * np.cos(2 * np.pi * n * f0 * (t + V / (4 * f0))), axis=0)
y_2 += +1 / 2 * np.sum(1 / n ** 2 * np.sin(n * np.pi / 2) * np.sin(2 * np.pi * n * f0 * (t - V / (4 * f0))), axis=0)
y_2 += +1 / 2 * np.sum(1 / n ** 2 * np.sin(n * np.pi / 2) * np.sin(2 * np.pi * n * f0 * (t + V / (4 * f0))), axis=0)

y_3 = np.zeros_like(y_0)
y_3 += +1 / 2 * np.sum((-1) ** n / (2 * n) ** 2 * np.cos(2 * np.pi * (2 * n) * f0 * (t - V / (4 * f0))), axis=0)
y_3 += -1 / 2 * np.sum((-1) ** n / (2 * n) ** 2 * np.cos(2 * np.pi * (2 * n) * f0 * (t + V / (4 * f0))), axis=0)
# y_3 += +1 / 2 * np.sum((-1) ** (n - 1) / (2 * n - 1) ** 2 * np.sin(2 * np.pi * (2 * n - 1) * f0 * (t - V / (4 * f0))), axis=0)
# y_3 += +1 / 2 * np.sum((-1) ** (n - 1) / (2 * n - 1) ** 2 * np.sin(2 * np.pi * (2 * n - 1) * f0 * (t + V / (4 * f0))), axis=0)

y_4 = np.zeros_like(y_0)
y_4 += + 1 / 8 * (2 * np.pi * f0 * t - np.pi * V / 2) ** 2 - np.pi ** 2 / 96
y_4 += - 1 / 8 * (2 * np.pi * f0 * t + np.pi * V / 2) ** 2 + np.pi ** 2 / 96

y_5 = np.zeros_like(y_0)
y_5 += np.pi / 4 * (2 * np.pi * f0 * t) * (1 - V)

y_6 = np.zeros_like(y_0)
y_6 += -np.pi / 4 * (2 * np.pi * f0 *t) * (1 + V)

plt.plot(2 * np.pi * f0 * t, y_0)
# plt.plot(2 * np.pi * f0 * t, y_1)
# plt.plot(2 * np.pi * f0 * t, y_3)
# plt.plot(2 * np.pi * f0 * t, y_4)
plt.plot(2 * np.pi * f0 * t, y_5)
plt.plot(2 * np.pi * f0 * t - np.pi, y_6)
plt.plot(2 * np.pi * f0 * t + np.pi, y_6)
plt.axvline(+ np.pi / 2 * (V - 1))
plt.axvline(+ np.pi / 2 * (V + 1))
plt.axvline(+ np.pi / 2 * (V - 1))
plt.axvline(+ np.pi / 2 * (V + 1))
plt.axvline(- np.pi / 2 * (V - 1))
plt.axvline(- np.pi / 2 * (V + 1))
plt.axvline(- np.pi / 2 * (V - 1))
plt.axvline(- np.pi / 2 * (V + 1))
```

```python
V = 0.4

t = np.arange(-1 / 2 / f0, 1 / 2 / f0, 1 / (1000 * f0))

theta_n = 2 * np.pi * f0 * t - np.pi * V / 2
theta_p = 2 * np.pi * f0 * t + np.pi * V / 2

y_c = np.zeros_like(t)
y_c -= np.sum((-1) ** n / (2 * n) ** 3 * np.sin((2 * n) * theta_n), axis=0)
y_c += np.sum((-1) ** n / (2 * n) ** 3 * np.sin((2 * n) * theta_p), axis=0)

y_d = np.zeros_like(t)
y_d += np.sum((-1) ** (n - 1) / (2 * n - 1) ** 3 * np.cos((2 * n - 1) * theta_n), axis=0)
y_d += np.sum((-1) ** (n - 1) / (2 * n - 1) ** 3 * np.cos((2 * n - 1) * theta_p), axis=0)

plt.plot(2 * np.pi * f0 * t, 1 / 2 * y_c)
plt.plot(2 * np.pi * f0 * t, 1 / 2 * y_d)
plt.plot(2 * np.pi * f0 * t, 1 / 2 * y_c +  1 / 2 * y_d)

plt.axvline(+ np.pi / 2 * (V - 1))
plt.axvline(- np.pi / 2 * (V - 1))
plt.axvline(+ np.pi / 2 * (V + 1))
plt.axvline(- np.pi / 2 * (V + 1))

plt.axhline(+np.pi / 2 * (1 - V))
plt.axhline(+np.pi / 2 * (1 + V))
plt.legend()
np.savetxt("filter-general.csv", np.column_stack((2 * np.pi * f0 * t, 1 / 2 * y_c, 1 / 2 * y_d, 1 / 2 * y_c +  1 / 2 * y_d)), delimiter=",")
```
